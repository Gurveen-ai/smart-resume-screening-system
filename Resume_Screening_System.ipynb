```python
import PyPDF2
import string
import nltk

from nltk.corpus import stopwords
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

nltk.download('stopwords')
```

```python
def extract_text_from_pdf(pdf_file):
    text = ""

    with open(pdf_file, 'rb') as file:
        reader = PyPDF2.PdfReader(file)

        for page in reader.pages:
            extracted = page.extract_text()
            if extracted:
                text += extracted

    return text
```

```python
resume_text = extract_text_from_pdf("Resume (1).pdf")
jd_text = extract_text_from_pdf("Job_Description.pdf")

print("Resume Text:")
print(resume_text[:500])

print("\nJob Description Text:")
print(jd_text[:500])
```

```python
def clean_text(text):
    text = str(text)
    text = text.lower()

    text = ''.join(char for char in text if char not in string.punctuation)

    words = text.split()

    stop_words = set(stopwords.words('english'))

    cleaned_words = [word for word in words if word not in stop_words and word.isalpha()]

    return ' '.join(cleaned_words)
```

```python
resume_clean = clean_text(resume_text)
jd_clean = clean_text(jd_text)

print(resume_clean[:300])
print(jd_clean[:300])
```

```python
resume_words = set(resume_clean.split())
jd_words = set(jd_clean.split())

matched_skills = resume_words.intersection(jd_words)

print("Matched Skills:")
print(matched_skills)
```

```python
missing_skills = jd_words - resume_words

print("Missing Skills:")
print(missing_skills)
```

```python
skills_list = [
    "python", "machine learning", "tensorflow", "keras",
    "scikit-learn", "pandas", "numpy", "sql",
    "deep learning", "cnn", "nlp", "matplotlib"
]
```

```python
resume_found_skills = []

for skill in skills_list:
    if skill in resume_clean:
        resume_found_skills.append(skill)

print("Skills Found in Resume:")
print(resume_found_skills)
```

```python
jd_found_skills = []

for skill in skills_list:
    if skill in jd_clean:
        jd_found_skills.append(skill)

print("Skills Required in JD:")
print(jd_found_skills)
```

```python
missing_real_skills = set(jd_found_skills) - set(resume_found_skills)

print("Missing Important Skills:")
print(missing_real_skills)
```
